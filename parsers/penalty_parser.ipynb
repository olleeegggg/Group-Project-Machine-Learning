## Сбор всех ударов и выделение среди них пенальти

Импортируем нужные библиотеки для работы с файловой системой, API и таблицами

```python
import os
import pandas as pd
import soccerdata as sd
import warnings
import itertools
from tqdm.auto import tqdm
import glob
import numpy as np
import shutil
import time
import sys

warnings.filterwarnings('ignore')
```

Создадим папку с таблицами, где каждая таблица - это один сезон в одной лиге. Всего 10 сезонов и 5 лиг, то есть суммарно 50 табличек.

Будем обрабатывать все удары, но выбирать из них только пенальти. Значение пенальти не хранится в множестве значений типа удара, поэтому будем обрабатывать пенальти чуть хитрее через координаты точки, с которой бьется пенальти.

```python
chunks_dir = os.path.join("..", "data", "penalty_raw_chunks")
```

```python
target_seasons = list(range(2016, 2026))
target_leagues = ['ENG-Premier League', 'ESP-La Liga', 'FRA-Ligue 1', 'GER-Bundesliga', 'ITA-Serie A']
tasks = list(itertools.product(target_leagues, target_seasons))
```

```python
os.makedirs(chunks_dir, exist_ok=True)

for league, season in tqdm(tasks, mininterval=0.1):
    file_name = f"raw_{league.replace(' ', '_')}_{season}.csv"
    chunk_file = os.path.join(chunks_dir, file_name)
    
    try:
        parser = sd.Understat(leagues=[league], seasons=[season])
        df_shots = parser.read_shot_events()
        
        if df_shots is not None and not df_shots.empty:
            df_shots = df_shots.reset_index()
            df_penalties = df_shots[
                (df_shots['location_x'].round(2) == 0.88) & 
                (df_shots['location_y'].round(2) == 0.50)
            ].copy()
            df_penalties['situation'] = 'Penalty'

            df_penalties.to_csv(chunk_file, index=False)
            
        #time.sleep(0.4)
        
    except Exception as e:
        tqdm.write(f"Пропущено {league} {season}: {e}")
        #time.sleep(1)

print("КОНЕЦ!!!!!!!!!!!!!!!!!!")
```

**NOTE:** Пропущено GER-Bundesliga 2024: 'list' object has no attribute 'values'

```python
# Ищем все CSV файлы в нашей папке
all_files = glob.glob(os.path.join(chunks_dir, "*.csv"))
df_raw = pd.concat((pd.read_csv(f) for f in all_files), ignore_index=True)

df_raw
```
